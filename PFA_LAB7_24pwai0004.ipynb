```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

**Task 1: Create a line graph showing how temperature changes
over 5 days.**

```python
days=['Sunday','Monday','Wednesday','Thursday','Friday']
Temperature=[33,43,32,30,30]
plt.plot(days,Temperature)
plt.title("Changes in temperature over the course of 5 days.")
plt.xlabel("Days")
plt.ylabel("Temperature in Celsius")
plt.grid(True)
plt.show()
```

**Task 2: Create a bar chart to compare marks of 4 students.**

```python
students=['Warisha','Maryam','Areej','Areeba']
marks=[85,89,90,99]
colors=['gray','pink','green','purple']
bars=plt.bar(students,marks,color=colors)
plt.title("Student marks comparison")
plt.bar_label(bars, padding=3)
plt.show()
```

**Task 3: Generate a histogram of 500 random exam scores.**

```python
scores=np.random.randint(0,100,500) #start range, end range, number of scores needed
plt.hist(scores, bins=10,color='pink',edgecolor='gray')
plt.title("Score distribution of 500 Exam Scores")
plt.xlabel("Score Range")
plt.ylabel("frequency")
plt.show()
```

**Task: Show relationship between study hours and marks.**

```python
hours=[1,2,3,4,5,6,7]
Marks=[40,45,50,60,70,80,90]
plt.scatter(hours, Marks, color='red')
plt.title("Marks obtained over number of hours studied")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.show()
```

**Task: Create a smooth line plot using Seaborn**
 X: [1,2,3,4,5]
 Y: [10,20,25,30,40]

```python
x=[1,2,3,4,5]
y=[10,20,25,30,40]
#Seaborn 
sns.lineplot(x=x,y=y)
plt.title("Seaborn Line Plot")
plt.show()
#matplotlib
plt.plot(x,y)
plt.title("matplotlib line plot")
plt.show()
```

**Task: Create a heatmap of a 5×5 random matrix.**

```python
matrix_data = np.random.rand(5, 5) 
sns.heatmap(matrix_data, annot=True, cbar=True)
plt.title("5x5 Random Matrix Heatmap") 
plt.show()
```

**Task: Plot same data using both libraries.**

```python
x_comp=[1,2,3,4,5]
y_comp=[2,4,6,8,10]
#Matplotlib Scatter
plt.scatter(x_comp,y_comp)
plt.title("Matplotlib Scatter")
plt.show()
#seaborn scatter
sns.scatterplot(x=x_comp,y=y_comp)
plt.title("seaborn scatter")
plt.show()
```

**Task: Create a mini dataset of your class (at least 5 students):**

```python
from matplotlib import markers
data={
    'Names':['Warisha','Areej','Rayyan','Maryam','Areeba'],
    'Marks':[79,89,76,90,76],
    'Attendance':[89,90,89,87,90]
}
df=pd.DataFrame(data)
#1. Make a bar chart for marks
plt.bar(df['Names'],df['Marks'],color='teal')
plt.title("Class Marks Distribution")
plt.show()
#2. Line Chart for Attendance
plt.plot(df['Names'],df['Attendance'],marker='s',color='blue')
plt.title("Attendance trend")
plt.show()
#3. Heatmap
correlation_matrix=df[['Marks','Attendance']].corr()
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm")
plt.title("Relation: Marks vs Attendance")
plt.show()
```
