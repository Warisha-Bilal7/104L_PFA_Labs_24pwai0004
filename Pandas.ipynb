Load customers.csv, purchases.csv and memberships.xlsx

```python
import pandas as pd
import numpy as np
from datetime import datetime

# 1. Load the Datasets
try:
    df_cust = pd.read_csv("customers.csv")
    df_purch = pd.read_csv("purchases.csv")
    df_memb = pd.read_excel("membership.xlsx", engine="openpyxl")
    print("✅ All datasets loaded successfully!")
except Exception as e:
    print(f"❌ Error loading files: {e}")

# 2. Previews (First 5 rows for space efficiency)
print("\n--- Customers Preview ---")
print(df_cust.head(10))

print("\n--- Purchases Preview ---")
print(df_purch.head(10))

print("\n--- Membership Preview ---")
print(df_memb.head(10))

# 3. Structural Properties
for name, df in [("Customers", df_cust), ("Purchases", df_purch), ("Membership", df_memb)]:
    print(f"\n=== Properties of {name} Dataset ===")
    print(f"Shape: {df.shape}")
    print("\nData Types:")
    print(df.dtypes)
    print("\nSummary Statistics:")
    print(df.describe(include='all'))
```
