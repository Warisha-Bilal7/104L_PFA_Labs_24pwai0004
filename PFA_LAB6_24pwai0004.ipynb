**Task 12: Data Normalization**

```python
print("12.1 Create an array of numerical data.")
data=np.array([0,50,100,150,200,250,300,350,400,450,500])
print("12.2. Normalize the dataset using formula:")
norm=(data-data.min()) / (data.max() -data.min())
print("12.3 Print normalized values.")
print(f"Normalized Data: {norm}")
print("12.4 Importance: Normalization ensures all features contribute equally and helps models converge faster during gradient descent.")
```
